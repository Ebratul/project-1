```python
for col in numeric_column:
  df[col]=df[col].fillna(df[col].median())
```

```python
for col in catagorical_column:
  df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
label_encoder = {}
for col in catagorical_column:
  le = LabelEncoder()
  df[col] = le.fit_transform(df[col])
  label_encoder[col] = le
```

```python
df
```

```python
label_encoder
```

```python
target_col = "Belief_Early_Hearing_Care"
if target_col not in df.columns:
  raise ValueError(f"Target Column {target_col} not found.please update targer_col variable.")
x = df.drop(columns=[target_col])
y = df[target_col]
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
val_scaller=scaler.fit_transform(x)
val_scaller
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(val_scaller, y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
```

```python
lg = LogisticRegression(max_iter=1000,class_weight='balanced')
lg.fit(x_train, y_train)
```

```python
print(classification_report(y_test, lg.predict(x_test)))
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
print(f"recall (macro):{recall_score(y_test,lg.predict(x_test), average='macro'):.4f}")
# print(f"recall (micro):{recall_score(y_test,lg.predict(x_test), average='micro')}")
```
